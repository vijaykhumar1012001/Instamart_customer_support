```python
import sqlite3
import pandas as pd
from pathlib import Path

# Connect to your SQLite database
conn = sqlite3.connect("data/instamart.db")

# Read the SQL file
sql_file = Path("sql/queries.sql")
sql_text = sql_file.read_text(encoding="utf-8")

# Split the file using the Q1, Q2, ... Q7 comment markers
sections = {}
current_question = None

for line in sql_text.splitlines():
    stripped = line.strip()

    if stripped.startswith("-- Q"):
        current_question = stripped.lstrip("- ").split(":")[0]
        sections[current_question] = []
    elif current_question is not None:
        sections[current_question].append(line)

# Execute each query and display the result
for question, lines in sections.items():
    query = "\n".join(lines).strip()

    if query:
        result = pd.read_sql_query(query, conn)

        print(f"\n{'=' * 40}")
        print(question)
        print(f"{'=' * 40}")
        print(result.to_string(index=False))
        print("Number of rows:", len(result))

conn.close()
```
